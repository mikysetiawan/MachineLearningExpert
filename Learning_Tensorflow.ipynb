REDUCING

* without axis = sum all
* axis 0 = sum column
* axis 1 = sum row

```python
tensor = tf.constant([[1, 2, 3], [4, 5, 6]])

sum_all = tf.reduce_sum(tensor)
sum_axis0 = tf.reduce_sum(tensor, axis=0)
sum_axis1 = tf.reduce_sum(tensor, axis=1)

print(tensor)
print(sum_all)
print(sum_axis0)
print(sum_axis1)
```

EAGER MODE VS GRAPH MODE

```python
def kalkulasi_rumit_eager(x):
  total = tf.constant(0.0)
  for i in tf.range(100000):
    total += tf.cast(i, dtype=tf.float32) * x
  return total


start_time = time.time()
hasil = kalkulasi_rumit_eager(tf.constant(2.0))
end_time = time.time()


print(f"Hasil Eager Mode: {hasil.numpy()}")
print(f"Waktu Eksekusi Eager Mode: {end_time - start_time:.4f} detik")
```

```python
@tf.function
def kalkulasi_rumit_graph(x):
  total = tf.constant(0.0)
  for i in tf.range(100000):
    total += tf.cast(i, dtype=tf.float32) * x
  return total

# Mengukur waktu eksekusi pertama kali
print("Eksekusi Pertama Graph (Kompilasi)")
start_time = time.time()
hasil = kalkulasi_rumit_graph(tf.constant(2.0))
end_time = time.time()

print(f"Hasil Graph Mode: {hasil.numpy()}")
print(f"Waktu Eksekusi Pertama Graph: {end_time - start_time:.4f} detik")

# Mengukur waktu eksekusi kedua
print("Eksekusi Kedua Graph")
start_time = time.time()
hasil = kalkulasi_rumit_graph(tf.constant(2.0))
end_time = time.time()

print(f"Hasil Graph Mode: {hasil.numpy()}")
print(f"Waktu Eksekusi Kedua Graph: {end_time - start_time:.4f} detik")
```

```python
# FUNCTIONAL API

from tensorflow.keras import Model
from tensorflow.keras.layers import Dense, Input

#Inisiasi input layer
input_layer = Input(shape=(4,))

#Susun layer yang terhubung hingga ke output layer (saling terhubung / sequental)
dense_1 = Dense(128, activation='relu')(input_layer)
dense_2 = Dense(64, activation='relu')(dense_1)
dense_3 = Dense(32, activation='relu')(dense_2)
dense_4 = Dense(16, activation='tanh')(dense_3)
output_layer = Dense(1, activation='sigmoid')(dense_4)

#Deklarasikan model dengan input dan output
model = Model(inputs=input_layer, outputs=output_layer)

print(model)

# ATAU dibuat paralel 1 to many
layer_1 = Dense(128, activation='relu')(input_layer)
layer_2 = Dense(32, activation="relu")(layer_1)
layer_3 = Dense(64, activation="relu")(layer_1)
layer_4 = Dense(16, activation="tanh")(layer_1)

print(layer_4)

# Bisa digabung lagi
from tensorflow.keras.layers import Concatenate
merge = Concatenate([layer_2, layer_3, layer_4])

print(merge)

# mutliple output
output_2 = Dense(1)(layer_2)
output_3 = Dense(1)(layer_3)
output_4 = Dense(1)(layer_4)
```

```python
# SIAMESE NETWORK, Multiple input

from tensorflow import keras
from tensorflow.keras.layers import Flatten, Dropout

def build_base_network(input_shape):
    input_layer = Input(shape=input_shape)
    x = Flatten()(input_layer)
    x = Dense(256, activation='relu')(x)
    x = Dropout(0.2)(x)
    x = Dense(128, activation='relu')(x)
    return Model(inputs=input_layer, outputs=x)

input_a = Input(shape=(28, 28,))
input_b = Input(shape=(28, 28,))


vec_a = build_base_network(input_a.shape[1:])
vec_b = build_base_network(input_b.shape[1:])

output = keras.ops.sqrt(keras.ops.sum(keras.ops.square(vec_a(input_a) - vec_b(input_b)), axis=1, keepdims=True))
siamese_net = Model(inputs=[input_a, input_b], outputs=output)

print(siamese_net)
```
