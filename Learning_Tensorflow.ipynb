```python
import tensorflow as tf

@tf.function
def my_function(x):
    print("Inside function:", x)
    return x

print(my_function(tf.constant(5)))
print(my_function(tf.constant(10)))
```

```python
import tensorflow as tf

a = tf.Variable(3)
b = tf.constant(3)

a.assign(5)
b.assign(5)

print("Nilai a:", a)
print("Nilai b:", b)
```
